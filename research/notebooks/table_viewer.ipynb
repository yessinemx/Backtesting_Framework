# Research Tables Viewer
This notebook loads every CSV table generated by the paper replication workflow from `research/outputs/tables`.
It uses `config.config_paper` so the lookup stays aligned with the repository configuration.

```python
import sys
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

cwd = Path.cwd().resolve()
candidate_roots = [cwd, *cwd.parents]
root_dir = next((path for path in candidate_roots if (path / "config").exists()), cwd)
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

from config import config_paper as research_config

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)

tables_dir = research_config.TABLES_DIR
table_paths = sorted(tables_dir.glob("*.csv"))
print(f"Tables directory: {tables_dir}")
print(f"CSV tables found: {len(table_paths)}")
table_paths
```

```python
if not table_paths:
    display(Markdown("**No CSV tables found in `research/outputs/tables`.**"))
else:
    for table_path in table_paths:
        frame = pd.read_csv(table_path)
        display(Markdown(f"## {table_path.stem}"))
        display(Markdown(f"Rows: {len(frame)} | Columns: {len(frame.columns)}"))
        display(frame)
```
